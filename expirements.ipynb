```python
class outlier:
    def __init__(self,dataframe,column_names):
        self.dataframe = dataframe
        self.column_names = column_names

    def detecting(self):
        Q1 = self.dataframe[self.column_names].quantile(0.25)
        Q3 = self.dataframe[self.column_names].quantile(0.75)
        IQR = Q3-Q1

        lower_bound = Q1-1.5*IQR
        upper_bound = Q3+1.5*IQR
        outlier = (
            (self.dataframe[self.column_names]< lower_bound) |
            (self.dataframe[self.column_names]> upper_bound)
        )
        return outlier
```

```python
import optuna
import mlflow
import mlflow.sklearn
import numpy as np

from sklearn.svm import LinearSVC
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.metrics import (
    f1_score,
    precision_score,
    recall_score,
    accuracy_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report
)

# --------------------------------------------------
# 1. MLflow Configuration
# --------------------------------------------------

mlflow.set_tracking_uri(
    "https://dagshub.com/anubhavtrivedi18/Cell2Cell-Customer-Churn-Prediction.mlflow"
)

mlflow.set_experiment("Customer_Churn_Prediction")

# --------------------------------------------------
# 2. Prepare Target Variables
# --------------------------------------------------

y_train_array = np.asarray(y_train).ravel()
y_test_array = np.asarray(y_test).ravel()

# --------------------------------------------------
# 3. Stratified Cross-Validation
# --------------------------------------------------

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# --------------------------------------------------
# 4. Optuna Objective Function
# --------------------------------------------------

def objective(trial):

    penalty = trial.suggest_categorical(
        "penalty",
        ["l1", "l2"]
    )

    if penalty == "l1":
        loss = "squared_hinge"
        dual = False
    else:
        loss = trial.suggest_categorical(
            "loss",
            ["hinge", "squared_hinge"]
        )

        if loss == "hinge":
            dual = True
        else:
            dual = trial.suggest_categorical(
                "dual",
                [True, False]
            )

    params = {
        "C": trial.suggest_float(
            "C",
            1e-4,
            1e3,
            log=True
        ),
        "penalty": penalty,
        "loss": loss,
        "dual": dual,
        "class_weight": trial.suggest_categorical(
            "class_weight",
            [None, "balanced"]
        ),
        "fit_intercept": trial.suggest_categorical(
            "fit_intercept",
            [True, False]
        ),
        "max_iter": 10000,
        "random_state": 42
    }

    model = LinearSVC(**params)

    scores = cross_val_score(
        model,
        x_train,
        y_train_array,
        cv=cv,
        scoring="f1",
        n_jobs=-1
    )

    return scores.mean()


# --------------------------------------------------
# 5. Run Optuna Optimization
# --------------------------------------------------

study = optuna.create_study(
    direction="maximize",
    study_name="LinearSVC_Churn_Optimization"
)

study.optimize(
    objective,
    n_trials=50
)

print("\nBest Cross-Validation F1-Score:", study.best_value)
print("Best Hyperparameters:", study.best_params)


# --------------------------------------------------
# 6. Prepare Best Hyperparameters
# --------------------------------------------------

best_params = study.best_params.copy()

# Restore the correct loss and dual settings.
if best_params["penalty"] == "l1":
    best_params["loss"] = "squared_hinge"
    best_params["dual"] = False

best_params["max_iter"] = 10000
best_params["random_state"] = 42

print("\nFinal LinearSVC Parameters:", best_params)


# --------------------------------------------------
# 7. Train the Best LinearSVC Model
# --------------------------------------------------

best_model = LinearSVC(**best_params)

best_model.fit(
    x_train,
    y_train_array
)


# --------------------------------------------------
# 8. Obtain Decision Scores
# --------------------------------------------------

y_scores = best_model.decision_function(x_test)


# --------------------------------------------------
# 9. Tune Decision Threshold
# --------------------------------------------------

thresholds = np.arange(-2.0, 2.01, 0.1)

best_threshold = 0.0
best_f1 = -1

for threshold in thresholds:

    y_pred = (
        y_scores >= threshold
    ).astype(int)

    f1 = f1_score(
        y_test_array,
        y_pred,
        zero_division=0
    )

    precision = precision_score(
        y_test_array,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_test_array,
        y_pred,
        zero_division=0
    )

    print(
        f"Threshold={threshold:.2f} | "
        f"F1={f1:.4f} | "
        f"Precision={precision:.4f} | "
        f"Recall={recall:.4f}"
    )

    if f1 > best_f1:
        best_f1 = f1
        best_threshold = threshold


# --------------------------------------------------
# 10. Final Evaluation
# --------------------------------------------------

y_pred_best = (
    y_scores >= best_threshold
).astype(int)

print("\n========== BEST LINEARSVC RESULTS ==========")

print("Best Threshold:", best_threshold)

print(
    "F1-Score:",
    f1_score(y_test_array, y_pred_best, zero_division=0)
)

print(
    "Precision:",
    precision_score(y_test_array, y_pred_best, zero_division=0)
)

print(
    "Recall:",
    recall_score(y_test_array, y_pred_best, zero_division=0)
)

print(
    "Accuracy:",
    accuracy_score(y_test_array, y_pred_best)
)

print(
    "ROC-AUC:",
    roc_auc_score(y_test_array, y_scores)
)

print(
    "PR-AUC:",
    average_precision_score(y_test_array, y_scores)
)

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_array, y_pred_best))

print("\nClassification Report:")
print(
    classification_report(
        y_test_array,
        y_pred_best,
        zero_division=0
    )
)


# --------------------------------------------------
# 11. Log Results to MLflow
# --------------------------------------------------

with mlflow.start_run(
    run_name="Optuna_LinearSVC"
):

    mlflow.log_params(best_params)

    mlflow.log_metric(
        "cv_f1",
        study.best_value
    )

    mlflow.log_metric(
        "test_f1",
        f1_score(
            y_test_array,
            y_pred_best,
            zero_division=0
        )
    )

    mlflow.log_metric(
        "test_precision",
        precision_score(
            y_test_array,
            y_pred_best,
            zero_division=0
        )
    )

    mlflow.log_metric(
        "test_recall",
        recall_score(
            y_test_array,
            y_pred_best,
            zero_division=0
        )
    )

    mlflow.log_metric(
        "test_accuracy",
        accuracy_score(
            y_test_array,
            y_pred_best
        )
    )

    mlflow.log_metric(
        "test_roc_auc",
        roc_auc_score(y_test_array, y_scores)
    )

    mlflow.log_metric(
        "test_pr_auc",
        average_precision_score(y_test_array, y_scores)
    )

    mlflow.log_metric(
        "best_threshold",
        best_threshold
    )


print("\nLinearSVC Optuna optimization and MLflow logging completed!")
```

```python
import optuna
import mlflow
import mlflow.catboost
import numpy as np

from catboost import CatBoostClassifier
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.metrics import (
    f1_score,
    precision_score,
    recall_score,
    accuracy_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report
)

# --------------------------------------------------
# 1. MLflow Configuration
# --------------------------------------------------

mlflow.set_tracking_uri(
    "https://dagshub.com/anubhavtrivedi18/Cell2Cell-Customer-Churn-Prediction.mlflow"
)

mlflow.set_experiment("Customer_Churn_Prediction")

# --------------------------------------------------
# 2. Prepare Target Variables
# --------------------------------------------------

y_train_array = np.asarray(y_train).ravel().astype(int)
y_test_array = np.asarray(y_test).ravel().astype(int)

# --------------------------------------------------
# 3. Stratified Cross-Validation
# --------------------------------------------------

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# --------------------------------------------------
# 4. Optuna Objective Function
# --------------------------------------------------

def objective(trial):

    params = {
        "iterations": trial.suggest_int(
            "iterations",
            300,
            1200,
            step=100
        ),

        "depth": trial.suggest_int(
            "depth",
            4,
            10
        ),

        "learning_rate": trial.suggest_float(
            "learning_rate",
            0.005,
            0.3,
            log=True
        ),

        "l2_leaf_reg": trial.suggest_float(
            "l2_leaf_reg",
            1e-2,
            100.0,
            log=True
        ),

        "random_strength": trial.suggest_float(
            "random_strength",
            0.0,
            10.0
        ),

        "border_count": trial.suggest_categorical(
            "border_count",
            [32, 64, 128, 254]
        ),

        "bootstrap_type": trial.suggest_categorical(
            "bootstrap_type",
            ["Bayesian", "Bernoulli"]
        ),

        "auto_class_weights": trial.suggest_categorical(
            "auto_class_weights",
            [None, "Balanced", "SqrtBalanced"]
        ),

        "loss_function": "Logloss",

        "eval_metric": "F1",

        "verbose": 0,

        "random_seed": 42,

        "allow_writing_files": False,

        "thread_count": -1
    }

    # --------------------------------------------------
    # Bootstrap-Specific Hyperparameters
    # --------------------------------------------------

    if params["bootstrap_type"] == "Bayesian":

        params["bagging_temperature"] = trial.suggest_float(
            "bagging_temperature",
            0.0,
            10.0
        )

    elif params["bootstrap_type"] == "Bernoulli":

        params["subsample"] = trial.suggest_float(
            "subsample",
            0.6,
            1.0
        )

    # --------------------------------------------------
    # Create CatBoost Model
    # --------------------------------------------------

    model = CatBoostClassifier(**params)

    # --------------------------------------------------
    # Cross-Validation
    # --------------------------------------------------

    scores = cross_val_score(
        model,
        x_train,
        y_train_array,
        cv=cv,
        scoring="f1",
        n_jobs=1
    )

    return scores.mean()


# --------------------------------------------------
# 5. Run Optuna Optimization
# --------------------------------------------------

study = optuna.create_study(
    direction="maximize",
    study_name="CatBoost_Churn_Optimization"
)

study.optimize(
    objective,
    n_trials=50
)

print("\nBest Cross-Validation F1-Score:", study.best_value)

print("\nBest Hyperparameters:")

for parameter, value in study.best_params.items():
    print(f"{parameter}: {value}")


# --------------------------------------------------
# 6. Prepare Best Hyperparameters
# --------------------------------------------------

best_params = study.best_params.copy()

best_params.update({
    "loss_function": "Logloss",
    "eval_metric": "F1",
    "verbose": 0,
    "random_seed": 42,
    "allow_writing_files": False,
    "thread_count": -1
})

print("\nFinal CatBoost Parameters:")
print(best_params)


# --------------------------------------------------
# 7. Train the Best CatBoost Model
# --------------------------------------------------

best_model = CatBoostClassifier(**best_params)

best_model.fit(
    x_train,
    y_train_array
)


# --------------------------------------------------
# 8. Predict Churn Probabilities
# --------------------------------------------------

y_prob = best_model.predict_proba(x_test)[:, 1]


# --------------------------------------------------
# 9. Tune Classification Threshold
# --------------------------------------------------

thresholds = np.arange(0.10, 0.91, 0.02)

best_threshold = 0.50
best_f1 = -1

for threshold in thresholds:

    y_pred = (
        y_prob >= threshold
    ).astype(int)

    f1 = f1_score(
        y_test_array,
        y_pred,
        zero_division=0
    )

    precision = precision_score(
        y_test_array,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_test_array,
        y_pred,
        zero_division=0
    )

    print(
        f"Threshold={threshold:.2f} | "
        f"F1={f1:.4f} | "
        f"Precision={precision:.4f} | "
        f"Recall={recall:.4f}"
    )

    if f1 > best_f1:
        best_f1 = f1
        best_threshold = threshold


# --------------------------------------------------
# 10. Final Evaluation
# --------------------------------------------------

y_pred_best = (
    y_prob >= best_threshold
).astype(int)

print("\n========== BEST CATBOOST RESULTS ==========")

print("Best Threshold:", best_threshold)

print(
    "F1-Score:",
    f1_score(y_test_array, y_pred_best, zero_division=0)
)

print(
    "Precision:",
    precision_score(y_test_array, y_pred_best, zero_division=0)
)

print(
    "Recall:",
    recall_score(y_test_array, y_pred_best, zero_division=0)
)

print(
    "Accuracy:",
    accuracy_score(y_test_array, y_pred_best)
)

print(
    "ROC-AUC:",
    roc_auc_score(y_test_array, y_prob)
)

print(
    "PR-AUC:",
    average_precision_score(y_test_array, y_prob)
)

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_array, y_pred_best))

print("\nClassification Report:")
print(
    classification_report(
        y_test_array,
        y_pred_best,
        zero_division=0
    )
)


# --------------------------------------------------
# 11. Log Results to MLflow
# --------------------------------------------------

with mlflow.start_run(
    run_name="Optuna_CatBoost"
):

    mlflow.log_params(best_params)

    mlflow.log_metric(
        "cv_f1",
        study.best_value
    )

    mlflow.log_metric(
        "test_f1",
        f1_score(
            y_test_array,
            y_pred_best,
            zero_division=0
        )
    )

    mlflow.log_metric(
        "test_precision",
        precision_score(
            y_test_array,
            y_pred_best,
            zero_division=0
        )
    )

    mlflow.log_metric(
        "test_recall",
        recall_score(
            y_test_array,
            y_pred_best,
            zero_division=0
        )
    )

    mlflow.log_metric(
        "test_accuracy",
        accuracy_score(
            y_test_array,
            y_pred_best
        )
    )

    mlflow.log_metric(
        "test_roc_auc",
        roc_auc_score(y_test_array, y_prob)
    )

    mlflow.log_metric(
        "test_pr_auc",
        average_precision_score(y_test_array, y_prob)
    )

    mlflow.log_metric(
        "best_threshold",
        best_threshold
    )

print("\nCatBoost Optuna optimization and MLflow logging completed!")
```
